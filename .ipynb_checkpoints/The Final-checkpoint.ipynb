```python
players = {
     'Josh Allen': {
        'position': 'QB',
        'team': 'BUF'
    },    
    'Bijan Robinson Jr': {
        'position':'RB',
        'team': 'ATL'
    },    
    'Jahmyr Gibbs': {
        'position': 'RB',
        'team': 'DET'
    },
    'Saquon Barkley': {
        'position': 'RB',
        'team': 'PHI'
    },
    'Christian McCaffrey': {
        'position': 'RB',
        'team': 'SF'
    },
    'CeeDee Lamb': {
        'position':'WR',
        'team': 'DAL'
    },
    'Puka Nacua': {
        'position':'WR',
        'team': 'LAR'
    },
    'Trey McBride': {
        'position':'TE',
        'team': 'ARI'
    },
    'Brandon Aubrey': {
        'position': 'K',
        'team': 'DAL'
    },
    'Denver Broncos': {
        'position': 'DEF',
        'team': 'DEN'
    }, 
}
players
```

```python
roster = {
    'QB': 'Josh Allen',
    'RB1': 'Bijan Robinson Jr',
    'RB2': 'Jahmyr Gibbs',
    'WR1': 'CeeDee Lamb', 
    'WR2': 'Puka Nacua',
    'TE': 'Trey McBride',
    'FLEX': 'Saquon Barkley',
    'K': 'Brandon Aubrey',
    'DEF': 'Denver Broncos'
}
roster
```

```python
bench = []
bench.append('Christian McCaffrey')
bench
```

```python
def flex_eligible(player_name):
    player = players[player_name]
    position = player['position']
    return position in ['RB', 'WR', 'TE']
```

```python
flex_eligible('Josh Allen') 
```

```python
def is_starting(player_name):
    return player_name in roster.values()
```

```python
is_starting('Saquon Barkley')
```

```python
def can_use_flex(player_name):
    return flex_eligible(player_name) and not is_starting(player_name)
```

```python
can_use_flex('Josh Allen')
```

```python
def add_to_bench(player_name):
    if player_name not in bench:
        bench.append(player_name)
```

```python
add_to_bench('Christian McCaffrey')
```

```python
def remove_from_bench(player_name):
    if player_name in bench:
        bench.remove(player_name)
```

```python
def get_starting_slot(player_name):
    for slot, player in roster.items():
        if player == player_name:
            return slot
    return None
```

```python
get_starting_slot('Saquon Barkley')
```

```python
def slot_eligible(player_name, slot):
    position = players[player_name]['position']
    
    if slot == 'QB':
        return position == 'QB'

    if slot == 'RB1' or slot == 'RB2':
        return position =='RB'

    if slot == 'WR1' or slot == 'WR2':
        return position == 'WR'

    if slot == 'TE':
        return position == 'TE'

    if slot == 'FLEX':
        return position in ['RB', 'WR', 'TE']

    if slot == 'K':
        return position == 'K' 

    if slot == 'DEF':
        return position == 'DEF'

    return False
```

```python
print(slot_eligible('Josh Allen', 'QB'))
print(slot_eligible('Christian McCaffrey', 'FLEX'))
print(slot_eligible('Josh Allen', 'FLEX'))
```

```python
def move_player(player_name, target_slot):
    
    if not is_owned(player_name):
        print("You don't own that player. You cheater!")
        return False

    old_slot = get_starting_slot(player_name)
    if not slot_eligible(player_name, target_slot):
        print("That player can't go in that slot.")
        return False 

    if old_slot is not None:
        roster[old_slot] = None

    current_player = roster.get(target_slot)

    if current_player is not None:
        if old_slot is not None:
            roster[old_slot] = current_player
        else:
            add_to_bench(current_player)

    remove_from_bench(player_name)
    roster[target_slot] = player_name
```

```python
move_player('Christian McCaffrey', 'RB1')
```

```python
print(roster)
print(bench) 
```

```python
def is_owned(player_name):
    return player_name in roster.values() or player_name in bench
```

```python
print(is_owned('Christian McCaffrey'))
print(is_owned('Josh Allen'))
```

```python
move_player('', 'FLEX')
```

```python
print(roster)
print(bench)
```

```python
def get_player_status(player_name):
    if player_name in roster.values():
        return "Starting"

    if player_name in bench:
        return 'Bench'

    return 'Not Owned'    
```

```python
print(get_player_status('Christian McCaffrey'))
print(get_player_status('Josh Allen'))
```
