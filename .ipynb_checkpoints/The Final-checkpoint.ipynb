```python
roster = {
    'QB': 'Josh Allen',
    'RB1': 'Bijan Robinson',
    'RB2': 'Jahmyr Gibbs',
    'WR1': 'CeeDee Lamb', 
    'WR2': 'Puka Nacua',
    'TE': 'Trey McBride',
    'FLEX': 'Saquon Barkley',
    'K': 'Brandon Aubrey',
    'DEF': 'Denver Broncos'
}
roster
```

```python
bench = []
bench.append('Christian McCaffrey')
bench
```

```python
def flex_eligible(player_name):
    player = players[player_name]
    position = player['position']
    return position in ['RB', 'WR', 'TE']
```

```python
flex_eligible('Josh Allen') 
```

```python
def is_starting(player_name):
    return player_name in roster.values()
```

```python
is_starting('Saquon Barkley')
```

```python
def can_use_flex(player_name):
    return flex_eligible(player_name) and not is_starting(player_name)
```

```python
can_use_flex('Josh Allen')
```

```python
def add_to_bench(player_name):
    if player_name not in bench:
        bench.append(player_name)
```

```python
add_to_bench('Christian McCaffrey')
```

```python
def remove_from_bench(player_name):
    if player_name in bench:
        bench.remove(player_name)
```

```python
def get_starting_slot(player_name):
    for slot, player in roster.items():
        if player == player_name:
            return slot
    return None
```

```python
get_starting_slot('Saquon Barkley')
```

```python
def slot_eligible(player_name, slot):
    position = players[player_name]['position']
    
    if slot == 'QB':
        return position == 'QB'

    if slot == 'RB1' or slot == 'RB2':
        return position =='RB'

    if slot == 'WR1' or slot == 'WR2':
        return position == 'WR'

    if slot == 'TE':
        return position == 'TE'

    if slot == 'FLEX':
        return position in ['RB', 'WR', 'TE']

    if slot == 'K':
        return position == 'K' 

    if slot == 'DEF':
        return position == 'DEF'

    return False
```

```python
print(slot_eligible('Josh Allen', 'QB'))
print(slot_eligible('Christian McCaffrey', 'FLEX'))
print(slot_eligible('Josh Allen', 'FLEX'))
```

```python
def move_player(player_name, target_slot):
    
    if not is_owned(player_name):
        print("You don't own that player. You cheater!")
        return False

    old_slot = get_starting_slot(player_name)
    if not slot_eligible(player_name, target_slot):
        print("That player can't go in that slot.")
        return False 

    if old_slot is not None:
        roster[old_slot] = None

    current_player = roster.get(target_slot)

    if current_player is not None:
        if old_slot is not None:
            roster[old_slot] = current_player
        else:
            add_to_bench(current_player)

    remove_from_bench(player_name)
    roster[target_slot] = player_name
```

```python
move_player('Christian McCaffrey', 'RB1')
```

```python
print(roster)
print(bench) 
```

```python
def is_owned(player_name):
    return player_name in roster.values() or player_name in bench
```

```python
def get_player_status(player_name):
    if player_name in roster.values():
        return "Starting"

    if player_name in bench:
        return 'Bench'

    return 'Not Owned'    
```

```python
scoring = {
    'passing_yard': 0.04,
    'passing_td': 4,
    'interception': -2, 
    'rushing_yard': 0.1,
    'rushing_td': 6,
    'reception' : 1,
    'receiving_yard': 0.1,
    'receiving_td': 6
}
```

```python
josh_stats = {
    'passing_yard': 250,
    'passing_td': 2,
    'interception': 1
}
```

```python
def calculate_fantasy_points(stats):
    points = 0

    for stat,value in stats.items():
        if stat in scoring:
            points += value * scoring[stat]

    return points
```

```python
print(calculate_fantasy_points(josh_stats))
```

```python
def get_player_points(player_name):

    if players[player_name]['position'] == 'DEF':
        team = players[player_name]['team']

        team_week4 = stats[
            (stats['team'] == team) &
            (stats['week'] == 4)
        ]

        defense_stats = team_week4[
            ['def_sacks', 'def_interceptions',
             'def_fumbles', 'def_tds', 'def_safeties']
        ].sum()

        return calculate_defense_points(defense_stats)

    real_stats = get_latest_stats(player_name)

    if players[player_name]['position'] == 'K':
        return real_stats['fg_made'] * 3 + real_stats['pat_made']

    fantasy_stats = convert_stats(real_stats)
    points = calculate_fantasy_points(fantasy_stats)

    return round(points, 2)
```

```python
print(get_player_points('Josh Allen'))
```

```python
def get_latest_stats(player_name):
    player = stats[stats['player_display_name'] == player_name] 
    return player.sort_values('week').iloc[-1]
```

```python
def convert_stats(player_stats):
    return {
        'passing_yard': player_stats['passing_yards'],
        'passing_td': player_stats['passing_tds'],
        'interception': player_stats['passing_interceptions'],
        'rushing_yard': player_stats['rushing_yards'],
        'rushing_td': player_stats['rushing_tds'],
        'reception': player_stats['receptions'],
        'receiving_yard': player_stats['receiving_yards'],
        'receiving_td': player_stats['receiving_tds']
    }
```

```python
def calculate_defense_points(stats):
    points = 0

    points += stats['def_sacks'] * 1
    points += stats['def_interceptions'] * 2
    points += stats['def_fumbles'] * 2
    points += stats['def_tds'] * 6
    points += stats['def_safeties'] * 2

    return points
```
